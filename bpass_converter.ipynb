```python
import os

from pathlib import Path
from matplotlib import pyplot as plt
plt.style.use(Path(os.environ["HOME"]) / "default.mplstyle")

from LWphotorates.stellar_spectra.utils import convert_bpass_spec_to_xr
```

```python
multiplicity = "bin"
metallicity = "zem5"

imf = "imf_chab100"
bpass_chab_ds = convert_bpass_spec_to_xr(imf, metallicity, multiplicity)

imf = "imf100_300"
bpass_th_ds = convert_bpass_spec_to_xr(imf, metallicity, multiplicity)

imf = "imf170_100"
bpass_bh_ds = convert_bpass_spec_to_xr(imf, metallicity, multiplicity)
```

```python
fig, ax = plt.subplots()
ax.plot(
    bpass_chab_ds.wavelength,
    bpass_chab_ds.luminosity.isel(age=3),
    lw=3, ls="-", c="C0")
ax.plot(
    bpass_chab_ds.wavelength,
    bpass_chab_ds.luminosity.isel(age=30),
    lw=3, ls="-", c="C1")
ax.plot(
    bpass_th_ds.wavelength,
    bpass_th_ds.luminosity.isel(age=3),
    lw=3, ls="--", c="C0")
ax.plot(
    bpass_th_ds.wavelength,
    bpass_th_ds.luminosity.isel(age=30),
    lw=3, ls="--", c="C1")
ax.plot(
    bpass_bh_ds.wavelength,
    bpass_bh_ds.luminosity.isel(age=3),
    lw=3, ls=":", c="C0")
ax.plot(
    bpass_bh_ds.wavelength,
    bpass_bh_ds.luminosity.isel(age=30),
    lw=3, ls=":", c="C1")
ax.set_yscale("log")
ax.set_xscale("log")
ax.set_ylim(bottom=1e32, top=1e42)
ax.set_xlim(1e2, 1e5)
ax.set_xlabel(r"$\lambda$  [$\AA$]")
ax.set_ylabel(r"L$_\lambda$  [erg s$^{-1}$ $\AA^{-1}$]")
```
